# Coordinated rounding in a rank-one block

Four midpoint residues have positive independent expected error. Because their sensitivities share one output direction, coordinated up/down choices can cancel exactly.

```python
import numpy as np

from experiments.discrepancy import (
    conditional_expectation_finish,
    enumerate_binary_optimum,
    null_space_reduce,
    same_grid_bernoulli_energy,
)
```

```python
block = np.ones((4, 1))
theta = np.full(4, 0.5)

_, exact_error = enumerate_binary_optimum(block, theta)
relaxed, trace = null_space_reduce(block, theta)
rounded, _ = conditional_expectation_finish(block, theta, relaxed)
coordinated_error = float(np.linalg.norm((rounded - theta) @ block) ** 2)
independent_reference = same_grid_bernoulli_energy(block, theta)

print('independent reference:', independent_reference)
print('residual coordinates:', trace['residual_size'])
print('rounded vector:', rounded.astype(int).tolist())
print('coordinated error:', coordinated_error)
```

```python
assert independent_reference == 1.0
assert trace['residual_size'] == 0
assert coordinated_error == exact_error == 0.0
assert int(rounded.sum()) == 2
```

The cancellation is exact only because this small example has a rank-one image and no fixed offsets. The full protocol separately tests higher rank, imbalance, clipping, and reuse.
