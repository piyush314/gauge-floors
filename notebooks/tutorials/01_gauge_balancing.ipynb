# Gauge balancing in two coordinates

A product-preserving inner transform can fix a badly scaled factorization without changing the matrix product. This example separates the factorization-energy gap from the remaining coordinate-range gap.

```python
import numpy as np

from experiments.nuclear import balance_transform, leading_metrics, transformed_factors
```

```python
A = np.diag([4.0, 0.25])
B = np.diag([0.25, 4.0])
T = balance_transform(A, B)
X, Y = transformed_factors(A, B, T=T)

before = leading_metrics(A, B)
after = leading_metrics(X, Y)
print('original objective/floor:', round(before['objective_floor_ratio'], 6))
print('balanced objective/floor:', round(after['objective_floor_ratio'], 6))
print('balanced energy/nuclear:', round(after['energy_nuclear_ratio'], 6))
print('transform:', np.round(T, 6).tolist())
```

```python
np.testing.assert_allclose(X @ Y, A @ B, atol=1e-14)
np.testing.assert_allclose(T, np.diag([0.25, 4.0]), atol=1e-14)
assert abs(after['energy_nuclear_ratio'] - 1.0) < 1e-12
assert abs(after['objective_floor_ratio'] - 2.0) < 1e-12
```

Balancing closes the factorization-energy gap exactly. The remaining factor of two is a range-to-energy effect; the larger experiments study flattening methods for that second gap.
